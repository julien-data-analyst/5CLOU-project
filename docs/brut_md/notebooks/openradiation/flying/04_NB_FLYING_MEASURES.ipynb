# 04 - CAPTURE DES MESURES AU VOL

Dans ce notebook, nous allons capturer les mesures au vol uniquement et ajouter d'autres informations.

```python
# Chargement des librairies
import pyspark.sql.functions as F
```

## Lecture de la table nettoyé `clean_measures`

```python
SILVER_TABLE_MEASURES = "abfss://0bcbec59-dbbb-41e1-a6a0-d3763e6c43e6@onelake.dfs.fabric.microsoft.com/23840c0a-00b7-4a0b-af79-896b6c9be935/Tables/silver/clean_measures"
```

```python
# Lecture de la table brute à nettoyer
df_measures = spark.read.format("delta").load(SILVER_TABLE_MEASURES)
```

## Filtrage sur les mesures au vol

```python
# DataFrame sur les mesures en vol
flying_measures = (df_measures
                        .filter("qualification = 'plane'")
                        .drop("aircraftType"))
```

## Ecriture de la table silver `flying_measures`

```python
flying_measures.write.format("delta").mode("overwrite").saveAsTable("silver.flying_measures")
```
