# 03 - INGESTION DES MESURES DANS LA TABLE RAW LAKEHOUSE

Dans ce notebook, on va lire et créer la table Delta concernant les mesures brutes.

```python
# Chargement des librairies
import requests
from pyspark.sql import SparkSession
import pyspark.sql.functions as F
import tarfile
from notebookutils import mssparkutils # La gestion du lakehouse
import os
from pathlib import Path
```

```python
def read_csv_file(spark, path):
    """
    Permets de lire un fichier CSV avec ";" comme séparation 
    et de le renvoyer en Spark DataFrame
    """
    df = (
            spark.read
            .option("header", True)
            .option("sep", ";")
            .option("inferSchema", True)
            .csv(path)
        )
    return df
```

## Lecture du fichier CSV des mesures de radiations relevées

```python
# Stockage du fichier correspondant
LAKEHOUSE_FILE = "Files/OpenRadiation/extracted/measurements_withoutEnclosedObject.csv"
```

```python
# Lecture du fichier CSV
df_measures = read_csv_file(spark, LAKEHOUSE_FILE)
```

```python
# Observation des colonnes 
df_measures.printSchema()
```

## Enregistrement en tant que table Delta "raw_measures"

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS raw")
```

```python
# Ecriture dans le lakehouse
df_measures.write.format("delta").mode("overwrite").saveAsTable("raw.radiation_measures")
```
